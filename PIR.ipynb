# Brownian Motion

```python
def Brownian_path(T, n): 
    dt = T/n
    W = np.zeros(n+1)
    for i in range(1,n+1): 
        W[i]=W[i-1]+np.random.normal(0,np.sqrt(dt)) 
    return W
```

```python
path=Brownian_path(100,1000)
plt.plot(path)
plt.show()
```
